## Що таке середовище

Уяви робоче місце для ремонту: щоб виконати роботу, потрібні інструменти й матеріали. Для програми теж потрібен набір засобів.

**Середовище Python** — це Python певної версії, доступні йому бібліотеки та налаштування. Бібліотека — набір готового коду для певних завдань.

**Інтерпретатор Python** — програма, яка виконує написані тобою команди Python.


## Зошит і комп’ютер, що виконує код

| Поняття | Що це означає |
|---|---|
| Зошит, або notebook (`.ipynb`) | Файл із поясненнями, клітинками коду та, якщо збережені, результатами |
| Середовище виконання, або runtime | Комп’ютер із пам’яттю, файлами та Python, який виконує клітинки |
| Середовище Python | Доступна цьому Python версія, бібліотеки та налаштування |

У звичайному хмарному режимі Colab команди виконуються на віртуальній машині Google. Браузер показує зошит і передає команди цій машині.

Збережений зошит може залишитися на Drive або GitHub, а runtime тимчасовий. Після його видалення змінні, встановлені в ньому бібліотеки та файли на його локальному диску потрібно створювати заново. Файли, збережені на Drive або GitHub, зберігаються окремо.


## Крок 1. Дізнаємося, який Python виконує клітинку

Запусти наступну клітинку. Вона тільки виводить інформацію: нічого не встановлює й не змінює файли.


```python
import sys

print("Урок 1: середовище Python")
print("Версія Python:", sys.version)
print("Файл Python:", sys.executable)
```

## Розбираємо кожен рядок

- `import sys` підключає стандартний модуль `sys` з інформацією про поточний Python. Він уже входить до Python.
- `print(...)` показує результат під клітинкою.
- Текст у лапках — підпис, який ми самі обрали.
- `sys.version` повертає відомості про версію Python.
- `sys.executable` повертає шлях до програми Python, яка виконує цю клітинку.
- Кома всередині `print` дозволяє вивести підпис і значення разом.

Номер версії та шлях можуть відрізнятися. Перепиши результат у чат: ми розберемо саме твоє середовище.
